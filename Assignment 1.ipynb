```python
# Create the given string
strM = "Python beginner tutorial"

# Convert the string to uppercase
print("Uppercase:", strM.upper())

# Convert the string to lowercase
print("Lowercase:", strM.lower())

# Capitalize the first character of the string
print("Capitalize:", strM.capitalize())

# Count the number of occurrences of character 't'
print("Count of 't':", strM.count('t'))

# Replace Python with Machine Learning
print("After replacement:", strM.replace("Python", "Machine Learning"))
```

```python
# Create the first tuple
tuple1 = (10, 20, 30)

# Create the second tuple
tuple2 = (40, 50, 60)

# Display both tuples
print("Tuple 1:", tuple1)
print("Tuple 2:", tuple2)
```

```python
# Concatenate the two tuples and store the result in t_combine
t_combine = tuple1 + tuple2

# Display the combined tuple
print("Combined tuple:", t_combine)
```

```python
# Repeat the elements of t_combine three times
t_repeat = t_combine * 3

# Display the repeated tuple
print("Repeated tuple:", t_repeat)
```

```python
# Access and print the 3rd element of t_combine
print("3rd element:", t_combine[2])
```

```python
# Access and print the first three elements of t_combine
print("First three elements:", t_combine[:3])
```

```python
# Access and print the last three elements of t_combine
print("Last three elements:", t_combine[-3:])
```
